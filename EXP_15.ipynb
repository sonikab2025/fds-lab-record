```python

import numpy as np
import pandas as pd

# Create dataset without CSV file

data = {
"Country": [
"France", "Spain", "Germany", "Spain", "Germany",
"France", "Spain", "France", np.nan, "France"
],
"Age": [
44.0, 27.0, 30.0, 38.0, 40.0,
35.0, np.nan, 48.0, 50.0, 37.0
],
"Salary": [
72000.0, 48000.0, 54000.0, 61000.0, np.nan,
58000.0, 52000.0, 79000.0, 83000.0, 67000.0
],
"Purchased": [
"No", "Yes", "No", "No", "Yes",
"Yes", "No", "Yes", "No", "Yes"
]
}

# Create DataFrame

df = pd.DataFrame(data)

# Display the dataset

print(df)
```

```python

print(df.info())
```

```python
print(df.Country.mode())
```

```python
print(df.Country.mode()[0])
```

```python

type(df.Country.mode())
```

```python

df['Country'] = df['Country'].fillna(df['Country'].mode()[0])
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Salary'] = df['Salary'].fillna(round(df['Salary'].mean()))
df
```

```python

pd.get_dummies(df.Country)
```

```python

updated_dataset=pd.concat([pd.get_dummies(df.Country),df.iloc[:,[1,2,3]]],axis=1)
updated_dataset
```

```python

updated_dataset['Purchased'] = updated_dataset['Purchased'].map({'No': 0, 'Yes': 1})
updated_dataset
```
